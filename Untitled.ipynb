```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from preprocessing import crop_frame, MASK_TOP_X


def inner_rectangle(frame, x_range=MASK_TOP_X):
    """Largest rectangle inside the octagon mask: full frame -> crop_frame (1350 x 1080)
    -> columns x 213-1136 (inclusive), all rows. 924 wide x 1080 high, no black corners."""
    x0, x1 = x_range
    return crop_frame(frame)[:, x0:x1 + 1]


def dhash64(frame, x_range=MASK_TOP_X):
    """64-bit difference hash (dHash, Ling et al. 2021) of one video frame.

    inner_rectangle -> grayscale -> shrink to 9 wide x 8 high (INTER_AREA) -> in each row,
    bit = 1 if a pixel is brighter than its left neighbour. 8 x 8 = 64 bits.
    Returns the hash as a Python int."""
    rect = inner_rectangle(frame, x_range)                       # 1080 x 924
    g = cv2.cvtColor(rect, cv2.COLOR_BGR2GRAY)
    small = cv2.resize(g, (9, 8), interpolation=cv2.INTER_AREA)  # 9 wide, 8 high
    bits = small[:, 1:] > small[:, :-1]                          # 8 x 8 booleans
    return int.from_bytes(np.packbits(bits.flatten()).tobytes(), "big")


def hamming(h1, h2):
    """Number of differing bits between two hashes (0 = identical, 64 = all differ)."""
    return (h1 ^ h2).bit_count()


def hash_hex(h):
    """Hash as a 16-character hex string, for printing and storing."""
    return f"{h:016x}"
```

```python
def read_frames_at(video_path, indices):
    """Read specific frames by sequential decoding (exact indices). Returns {index: full BGR frame}."""
    wanted = set(int(i) for i in indices)
    out, i = {}, 0
    cap = cv2.VideoCapture(video_path)
    while wanted - out.keys():
        ret, frame = cap.read()
        if not ret:
            break
        if i in wanted:
            out[i] = frame
        i += 1
    cap.release()
    return out


video_path = r"C:\Users\ahmed\Documents\Jobs\Klinikum Oldenburg\Selected_videos\OA2459C_05-05-26_102237.mp4"
start = 600                                          # a frame inside a kept segment (580-854)
idx = list(range(start, start + 41, 2)) + [start + 50, start + 100, start + 200]   # even frames only (Decision 14)
frames = read_frames_at(video_path, idx)
test_hashes = {i: dhash64(frames[i]) for i in idx}

print(f"frame {start} vs itself: {hamming(test_hashes[start], test_hashes[start])} bits\n")
print(f"{'frame':>6} {'hash':>18} {'dist to prev':>13} {'dist to frame ' + str(start):>18}")
prev = None
for i in idx:
    d_prev = "" if prev is None else hamming(test_hashes[i], test_hashes[prev])
    print(f"{i:>6} {hash_hex(test_hashes[i]):>18} {str(d_prev):>13} {hamming(test_hashes[i], test_hashes[start]):>18}")
    prev = i
```

```python
def show_dhash(frame, x_range=MASK_TOP_X, title=""):
    """What the hash 'sees': the rectangle, its 9x8 shrunk version, and the 64 bits."""
    x0, x1 = x_range
    rect = crop_frame(frame)[:, x0:x1 + 1]
    g = cv2.cvtColor(rect, cv2.COLOR_BGR2GRAY)
    small = cv2.resize(g, (9, 8), interpolation=cv2.INTER_AREA)
    bits = small[:, 1:] > small[:, :-1]
    fig, ax = plt.subplots(1, 3, figsize=(15, 4.5))
    ax[0].imshow(cv2.cvtColor(rect, cv2.COLOR_BGR2RGB)); ax[0].set_title(f"{title} rectangle (924 x 1080)")
    ax[1].imshow(small, cmap="gray", interpolation="nearest"); ax[1].set_title("shrunk to 9 x 8 (grayscale)")
    ax[2].imshow(bits, cmap="gray", interpolation="nearest", vmin=0, vmax=1)
    ax[2].set_title("64 bits (white = brighter than left neighbour)")
    for a in ax: a.axis("off")
    plt.tight_layout(); plt.show()

show_dhash(frames[start], title=f"frame {start}")
show_dhash(frames[start + 200], title=f"frame {start + 200}")
```
